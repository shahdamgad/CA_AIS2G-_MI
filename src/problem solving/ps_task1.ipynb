### 13. Roman to Integer

Hint
Roman numerals are represented by seven different symbols: I, V, X, L, C, D and M.

Symbol       Value
I             1
V             5
X             10
L             50
C             100
D             500
M             1000
For example, 2 is written as II in Roman numeral, just two ones added together. 12 is written as XII, which is simply X + II. The number 27 is written as XXVII, which is XX + V + II.

Roman numerals are usually written largest to smallest from left to right. However, the numeral for four is not IIII. Instead, the number four is written as IV. Because the one is before the five we subtract it making four. The same principle applies to the number nine, which is written as IX. There are six instances where subtraction is used:

I can be placed before V (5) and X (10) to make 4 and 9. 
X can be placed before L (50) and C (100) to make 40 and 90. 
C can be placed before D (500) and M (1000) to make 400 and 900.
Given a roman numeral, convert it to an integer.

 

Example 1:

Input: s = "III"
Output: 3
Explanation: III = 3.
Example 2:

Input: s = "LVIII"
Output: 58
Explanation: L = 50, V= 5, III = 3.
Example 3:

Input: s = "MCMXCIV"
Output: 1994
Explanation: M = 1000, CM = 900, XC = 90 and IV = 4.
 

Constraints:

1 <= s.length <= 15
s contains only the characters ('I', 'V', 'X', 'L', 'C', 'D', 'M').
It is guaranteed that s is a valid roman numeral in the range [1, 3999]

```python
class Solution:
    def romanToInt(self, s: str) -> int:
        """
        Convert a Roman numeral string into an integer.

        Args:
            s (str): A valid Roman numeral.

        Returns:
            int: The integer value of the Roman numeral.
        """

        # Map each Roman numeral symbol to its integer value
        values = {
            'I': 1,
            'V': 5,
            'X': 10,
            'L': 50,
            'C': 100,
            'D': 500,
            'M': 1000
        }

        result: int = 0

        # Go through each character in the Roman numeral
        for i in range(len(s)):

            # If the current value is smaller than the next value,
            # subtract it (e.g., IV = 4, IX = 9)
            if i + 1 < len(s) and values[s[i]] < values[s[i + 1]]:
                result -= values[s[i]]

            # Otherwise, add the current value
            else:
                result += values[s[i]]

        return result


if __name__ == "__main__":

 solution = Solution()

 # Example 1
 s1 = "III"
 print("Example 1:", solution.romanToInt(s1))


 # Example 2
 s2 = "LVIII"
 print("Example 2:", solution.romanToInt(s2))


 # Example 3
 s3 = "MCMXCIV"
 print("Example 3:", solution.romanToInt(s3))
```

### 35. Search Insert Position

Given a sorted array of distinct integers and a target value, return the index if the target is found. If not, return the index where it would be if it were inserted in order.

You must write an algorithm with O(log n) runtime complexity.


Example 1:

Input: nums = [1,3,5,6], target = 5
Output: 2
Example 2:

Input: nums = [1,3,5,6], target = 2
Output: 1
Example 3:

Input: nums = [1,3,5,6], target = 7
Output: 4
 

Constraints:

1 <= nums.length <= 104
-104 <= nums[i] <= 104
nums contains distinct values sorted in ascending order.
-104 <= target <= 104

```python
from typing import List

class Solution:
    def searchInsert(self, nums: List[int], target: int) -> int:
        """
        Find the index of target in a sorted array.
        If target is not found, return the index where it should be inserted.

        Args:
            nums (List[int]): A sorted list of distinct integers.
            target (int): The value to search for.

        Returns:
            int: The index of target or its correct insertion position.
        """

        # Set the left and right boundaries
        left: int = 0
        right: int = len(nums) - 1

        # Binary search
        while left <= right:
            # Find the middle index
            mid: int = (left + right) // 2

            # Target found
            if nums[mid] == target:
                return mid

            # Target is greater, search in the right half
            elif nums[mid] < target:
                left = mid + 1

            # Target is smaller, search in the left half
            else:
                right = mid - 1

        # If target is not found, left is the insertion position
        return left


if __name__ == "__main__":

 solution = Solution()

 # Example 1
 nums1 = [1, 3, 5, 6]
 target1 = 5

 print("Example 1:", solution.searchInsert(nums1, target1))
 

 # Example 2
 nums2 = [1, 3, 5, 6]
 target2 = 2

 print("Example 2:", solution.searchInsert(nums2, target2))
 


 # Example 3
 nums3 = [1, 3, 5, 6]
 target3 = 7

 print("Example 3:", solution.searchInsert(nums3, target3))
```

### 227. Basic Calculator II
Given a string s which represents an expression, evaluate this expression and return its value. 

The integer division should truncate toward zero.

You may assume that the given expression is always valid. All intermediate results will be in the range of [-231, 231 - 1].

Note: You are not allowed to use any built-in function which evaluates strings as mathematical expressions, such as eval().

 

Example 1:

Input: s = "3+2*2"
Output: 7
Example 2:

Input: s = " 3/2 "
Output: 1
Example 3:

Input: s = " 3+5 / 2 "
Output: 5
 

Constraints:

1 <= s.length <= 3 * 105
s consists of integers and operators ('+', '-', '*', '/') separated by some number of spaces.
s represents a valid expression.
All the integers in the expression are non-negative integers in the range [0, 231 - 1].
The answer is guaranteed to fit in a 32-bit integer.

```python
class Solution:
    def calculate(self, s: str) -> int:
        """
        Evaluate a basic mathematical expression containing
        +, -, *, and / operators.

        Args:
            s (str): A valid mathematical expression.

        Returns:
            int: The result of the expression.
        """

        # Stack stores numbers that will be added together at the end
        stack = []

        # Current number
        num: int = 0

        # Operator before the current number
        operator: str = '+'

        # Add '+' at the end to process the last number
        for char in s + '+':

            # Build the number
            if char.isdigit():
                num = num * 10 + int(char)

            # Process the number when we reach an operator
            elif char in "+-*/":

                if operator == '+':
                    stack.append(num)

                elif operator == '-':
                    stack.append(-num)

                elif operator == '*':
                    stack[-1] *= num

                elif operator == '/':
                    # Truncate division toward zero
                    stack[-1] = int(stack[-1] / num)

                # Save the current operator
                operator = char

                # Reset number
                num = 0

        return sum(stack)



if __name__ == "__main__":
 solution = Solution()

 # Example 1
 s1 = "3+2*2"
 print("Example 1:", solution.calculate(s1))

 # Example 2
 s2 = " 3/2 "
 print("Example 2:", solution.calculate(s2))

 # Example 3
 s3 = " 3+5 / 2 "
 print("Example 3:", solution.calculate(s3))
 
```
