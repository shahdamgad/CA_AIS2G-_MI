```python
class BankAccount:

    def __init__(self, initial_balance):
        self.balance = initial_balance

    def deposit(self, amount):
        if amount < 0:
            print("Deposit amount cannot be negative.")
        else:
            self.balance += amount
            print(f"Deposited {amount}. New balance: {self.balance}")

    def withdraw(self, amount):
        if amount < 0:
            print("Withdrawal amount cannot be negative.")
        elif amount > self.balance:
            print("Insufficient funds")
        else:
            self.balance -= amount
            print(f"Withdrew {amount}. New balance: {self.balance}")

    def check_balance(self):
        print(f"Current balance: {self.balance}")


# Test
account = BankAccount(100)

account.deposit(50)
account.withdraw(30)
account.check_balance()
```

```python
class Calculator:
    def __init__(self):
        """Initializes the calculator class."""
        pass

    def add(self, a, b):
        """Add two numbers."""
        return a + b

    def subtract(self, a, b):
        """Subtract two numbers."""
        return a - b

    def multiply(self, a, b):
        """Multiply two numbers."""
        return a * b

    def divide(self, a, b):
        """Divide two numbers."""
        if b != 0:
            return a / b
        else:
            return "Error! Division by zero."

    def calculator(self):
        """Main calculator logic to get user input and perform operations."""
        print("Welcome to the calculator!")
        print("Choose an operation:")
        print("1: Add")
        print("2: Subtract")
        print("3: Multiply")
        print("4: Divide")

        choice = input(
            "Enter the number of the operation you want to perform (1/2/3/4): "
        )

        if choice in ['1', '2', '3', '4']:
            try:
                num1 = float(input("Enter the first number: "))
                num2 = float(input("Enter the second number: "))
            except ValueError:
                print("Invalid input! Please enter numbers only.")
                return

            if choice == '1':
                print(f"{num1} + {num2} = {self.add(num1, num2)}")

            elif choice == '2':
                print(f"{num1} - {num2} = {self.subtract(num1, num2)}")

            elif choice == '3':
                print(f"{num1} * {num2} = {self.multiply(num1, num2)}")

            elif choice == '4':
                result = self.divide(num1, num2)

                if isinstance(result, str):
                    print(result)
                else:
                    print(f"{num1} / {num2} = {result}")

        else:
            print("Invalid choice! Please select a valid operation (1/2/3/4).")


# Usage example
if __name__ == "__main__":
    calculator = Calculator()
    calculator.calculator()
```

```python
from abc import ABC, abstractmethod


class Animal(ABC):

    @abstractmethod
    def make_sound(self):
        pass

    def describe(self):
        print("This is an animal.")


class Dog(Animal):

    def make_sound(self):
        return "Woof!"


class Cat(Animal):

    def make_sound(self):
        return "Meow!"


class Cow(Animal):

    def make_sound(self):
        return "Moo!"


# Test the classes

dog = Dog()
cat = Cat()
cow = Cow()

print("Dog:")
print("Sound:", dog.make_sound())
dog.describe()

print("\nCat:")
print("Sound:", cat.make_sound())
cat.describe()

print("\nCow:")
print("Sound:", cow.make_sound())
cow.describe()
```

```python
class TextFileReader:

    def __init__(self, file_path):
        self.file_path = file_path
        self.content = ""

    def read_file(self):
        with open(self.file_path, "r") as file:
            self.content = file.read()

    def count_lines(self):
        return len(self.content.splitlines())

    def count_words(self):
        return len(self.content.split())

    def count_characters(self):
        return len(self.content)

    def display_content(self):
        print(self.content)


# Test
reader = TextFileReader("example.txt")
with open("example.txt", "w") as file:
    file.write("""Hello, my name is Shahd.
I am learning Python.
This is an OOP task.""")
    
reader.read_file()

print("Number of lines:", reader.count_lines())
print("Number of words:", reader.count_words())
print("Number of characters:", reader.count_characters())

print("\nFile Content:")
reader.display_content()
```

```python
def count_unique_words(words):
    unique_words = set(words)
    counts = {word: 0 for word in unique_words}

    for i in unique_words:
        for j in words:
            if i == j:
                counts[i] += 1

    return counts


# Test
words = ["Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"]

result = count_unique_words(words)

print(result)
```

```python
class WordCounter:

    def __init__(self, words):
        self.words = words

    def count_unique_words(self):
        unique_words = set(self.words)
        counts = {word: 0 for word in unique_words}

        for i in unique_words:
            for j in self.words:
                if i == j:
                    counts[i] += 1

        return counts


# Test
words = ["Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"]

counter = WordCounter(words)

print(counter.count_unique_words())
```

```python
# Create a test text file
with open("test.txt", "w", encoding="utf-8") as file:
    file.write("""Ali:12345
Ahmed:abc123
Mona:password1
Shahd:python123
""")
    

def read_txt_file(file_path):
    """Reads the contents of a text file and returns it as a string."""
    try:
        with open(file_path, 'r', encoding='utf-8') as file:
            content = file.read()
        return content

    except FileNotFoundError:
        return f"Error: The file '{file_path}' was not found."

    except IOError:
        return "Error: An error occurred while reading the file."


class UserExtractor:

    def __init__(self, file_path):
        # Initialize the class with file_path and an empty dictionary
        self.file_path = file_path
        self.usernames = {}

    def extract_usernames(self):
        """Extracts usernames from the text file and stores them in a dictionary."""

        # Read the file using the external function
        content = read_txt_file(self.file_path)

        # Check if an error occurred
        if content.startswith("Error:"):
            return content

        # Process each line
        for line in content.splitlines():

            # Remove extra spaces
            line = line.strip()

            # Skip empty lines
            if not line:
                continue

            # Extract username and password
            username, password = line.split(":", 1)

            # Store username and password
            self.usernames[username] = password

        return self.usernames


# Usage example
if __name__ == "__main__":

    file_path = "test.txt"

    extractor = UserExtractor(file_path)

    result = extractor.extract_usernames()

    print(result)
```
