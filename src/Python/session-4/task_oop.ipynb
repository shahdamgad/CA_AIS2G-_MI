```python
# problem 4
'''def dec_to_bin(n: int) -> int:
    """
    Convert a decimal number to its binary representation.

    Args:
        n (int): The decimal number to convert.

    Returns:
        int: The binary representation of the number.

    Example:
        dec_to_bin(11)
        1011
    """
    if n == 0:
        return 0

    binary = ""

    while n > 0:
        remainder = n % 2
        binary = str(remainder) + binary
        n //= 2

    return int(binary)


if __name__ == "__main__":
    print(dec_to_bin(11))'''

class DecimalToBinary:
    """
    A class for converting a decimal number to binary.
    """

    def __init__(self, number: int):
        """
        Initialize the DecimalToBinary object.

        Args:
            number (int): The decimal number to convert.
        """
        self.number = number

    def convert(self) -> int:
        """
        Convert the stored decimal number to binary.

        Returns:
            int: The binary representation of the number.
        """
        if self.number == 0:
            return 0

        n = self.number
        binary = ""

        while n > 0:
            remainder = n % 2
            binary = str(remainder) + binary
            n //= 2

        return int(binary)


if __name__ == "__main__":
    number = DecimalToBinary(11)
    print(number.convert())
```

```python
# problem 1
'''def factorial(num : int) -> int:
     calculate n! using recursion 
    Args:
        num (int): user input the int number
    Returns:
       num! (int): factorial of the number
    Example:
        factorial(5) returns 120
    

    if num == 0:
        return 1
    else:
        return num * factorial(num - 1)'''

class Factorial:
    """
    A class for calculating factorial using recursion.
    """

    def __init__(self, num: int):
        """
        Initialize the Factorial object.

        Args:
            num (int): The number to calculate the factorial for.
        """
        self.num = num

    def calculate(self, num=None) -> int:
        """
        Calculate factorial recursively.

        Args:
            num (int, optional): The number used in the recursive calculation.

        Returns:
            int: The factorial of the number.
        """
        if num is None:
            num = self.num

        if num == 0:
            return 1

        return num * self.calculate(num - 1)


if __name__ == "__main__":
    factorial = Factorial(5)
    print(factorial.calculate())
```

```python
# problem 3 
''' def common_divisor(num1 : int, num2 : int ) -> list[int]:
    
    function help to calc the common divisor of two numbers

    limit = min(num1, num2)
    divs = []

    for divisor in range(1, limit + 1):
        if num1 % divisor == 0 and num2 % divisor == 0:
            divs.append(divisor)
    return divs '''

class CommonDivisor:
    """
    A class for finding the common divisors of two numbers.
    """

    def __init__(self, num1: int, num2: int):
        """
        Initialize the CommonDivisor object.

        Args:
            num1 (int): The first number.
            num2 (int): The second number.
        """
        self.num1 = num1
        self.num2 = num2

    def find_common_divisors(self) -> list[int]:
        """
        Find all common divisors of the two stored numbers.

        Returns:
            list[int]: A list containing the common divisors.
        """
        limit = min(self.num1, self.num2)
        divs = []

        for divisor in range(1, limit + 1):
            if self.num1 % divisor == 0 and self.num2 % divisor == 0:
                divs.append(divisor)

        return divs


if __name__ == "__main__":
    numbers = CommonDivisor(12, 18)
    print(numbers.find_common_divisors())
```

```python
# problem 4
'''words = ["Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"]
def count_unique_words(words : list) -> dict:
 # find unique words in the list
 unique_words = set(words)
 result = " "
 for word in unique_words:
    result += f"{word.lower()}:{words.count(word)}\n"

 return(result)'''

class WordCounter:
    """
    A class for counting the occurrences of unique words in a list.
    """

    def __init__(self, words: list):
        """
        Initialize the WordCounter object.

        Args:
            words (list): A list of words to count.
        """
        self.words = words

    def count_unique_words(self) -> str:
        """
        Count how many times each unique word appears in the list.

        Returns:
            str: A string containing each unique word and its count.
        """
        unique_words = set(self.words)
        result = ""

        for word in unique_words:
            result += f"{word.lower()}:{self.words.count(word)}\n"

        return result


if __name__ == "__main__":
    words = [
        "Welcome", "Ali", "Hi", "Ali", "No",
        "Hi", "No", "Ali", "No", "Ali"
    ]

    counter = WordCounter(words)
    print(counter.count_unique_words())
```
