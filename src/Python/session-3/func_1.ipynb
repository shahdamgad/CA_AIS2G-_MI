```python
def info_user(name : str="shahd",age : int | None = None ) -> list:    # this is type hint :str   :int , default argument 'amit' 10
    # this is docstring
    '''   this function will return the name and age of the user          
    Args:
    name(str):name of the user
    age(int): age of the user
    Returns:
      str: name and age of the user

    '''
    return f"my name is {name} and i am {age} years old"
```

```python
info_user()            # write docstring and type hint for all functions
```

```python
def func(item,lst=[]):
    lst.append(item)
    return lst

print(func("1"))
print(func("2"))
print(func("3"))
```

```python
def func(item,lst=None):             # don't use mutable objects like list,dict instead Use none then create 
    if lst is None:                  
        lst = []

        lst.append(item)
        return lst

print(func("1"))
print(func("2"))
print(func("3"))
```

```python
def train_model(model_name,*datasets,**hyperparameters):
    print(f"training model:{model_name}")

    print("\n Datasets")
    for dataset in datasets:
        print("-", dataset)

    print("\n Hyperparameters")
    for key,value in hyperparameters.items():
        print(f"{key} : {value}")
```

```python
train_model(
   "fraud detection",
   "train.csv",
    "validation.csv",
    "test.csv",
    learning_rate=0.001,
    epochs=100,
    batch_size =64,
    optimizer="Adam",
    early_stopping=True

)
   
```
