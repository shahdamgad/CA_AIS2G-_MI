```python
def array_reshape(arr , rows , cols):
    try:
        arr = np.array(arr)       # convert lst to array  as v1=np.array(lst)
        return arr.reshape(rows,cols)
    
    except(TypeError,ValueError):
        raise ValueError(f" invalid reshape operation:{e}")
```

```python
vec = [1,2,3,4,5,6]
array_reshape(vec,2,3)
```

```python
def convert_dtype(ar, dtype):
    try:
        arr = np.array(arr)
        return arr.astype(dtype)
    except TypeError as e:
        raise TypeError(f" invalid target data type:{dtype}") from e

   
```
